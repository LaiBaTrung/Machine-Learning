```python
#14
missing = df.isna().sum()
print(missing)
print("Tổng số ô thiếu:", missing.sum())
```

```python
#15
n_dup = df.duplicated().sum()
print("Số bản ghi trùng lặp:", n_dup)
if n_dup > 0:
    print(df[df.duplicated(keep=False)].sort_values(by=df.columns.tolist()).head(10))
```

```python
#16
def iqr_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    n_out = ((series < lower) | (series > upper)).sum()
    return lower, upper, n_out

for col in ['MedInc', 'AveRooms', 'Population', 'AveOccup']:
    lower, upper, n_out = iqr_outliers(df[col])
    print(f"{col}: lower={lower:.2f}, upper={upper:.2f}, số outlier={n_out} ({n_out/len(df)*100:.2f}%)")
fig, axes = plt.subplots(1, 4, figsize=(16, 5))
for ax, col in zip(axes, ['MedInc', 'AveRooms', 'Population', 'AveOccup']):
    ax.boxplot(df[col])
    ax.set_title(f'Boxplot: {col}')
    ax.set_ylabel(col)
plt.tight_layout()
plt.show()
```

# Câu 4. Xây dựng mô hình Linear Regression (10 điểm)

```python
#21
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

print(X_train.dtypes)
```

```python
#22
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)

coef_df = pd.DataFrame({
    'Feature': X_train.columns,
    'Coefficient': model.coef_
})
coef_df['abs_coef'] = coef_df['Coefficient'].abs()
coef_df = coef_df.sort_values(by='abs_coef', ascending=False).drop(columns='abs_coef').reset_index(drop=True)
print(coef_df)
```

```python
#23
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)
```

```python
#24
result_df = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_test_pred,
    'Absolute Error': (y_test.values - y_test_pred)
})
result_df['Absolute Error'] = result_df['Absolute Error'].abs()
print(result_df.sample(10, random_state=42))
```

- intercept_ là giá trị dự đoán của MedHouseVal khi toàn bộ feature bằng 0 — đây là một điểm neo toán học của phương trình hồi quy, thường không có ý nghĩa thực tế trực tiếp (vì không khu vực nào có MedInc = 0, Latitude = 0... đồng thời).
- Mỗi coef_ cho biết: khi feature đó tăng thêm 1 đơn vị, MedHouseVal dự kiến thay đổi bao nhiêu đơn vị, với điều kiện giữ nguyên tất cả các feature khác (ceteris paribus) — đây là cách diễn giải chuẩn của hồi quy tuyến tính đa biến. - Hệ số dương (ví dụ MedInc) gợi ý feature đó tăng thì giá nhà trung vị dự kiến tăng theo; hệ số âm (ví dụ Latitude, Longitude) gợi ý chiều ngược lại.

- R² = 1 là mô hình dự đoán hoàn hảo; R² = 0 nghĩa là mô hình không tốt hơn việc luôn dự đoán bằng giá trị trung bình của y. R² có thể ÂM nếu mô hình dự đoán tệ hơn cả việc luôn đoán giá trị trung bình — điều này thường xảy ra khi mô hình được đánh giá trên tập test có phân bố khác biệt lớn so với tập train, hoặc mô hình bị overfitting nghiêm trọng.

#Nhận xét:
#So sánh Train vs Test — dấu hiệu underfitting/overfitting:

- Nếu R²(train) và R²(test) đều thấp và gần nhau → mô hình underfitting (mô hình quá đơn giản, không nắm bắt đủ quan hệ trong dữ liệu — hồi quy tuyến tính vốn có giới hạn này).
- Nếu R²(train) cao hơn đáng kể so với R²(test) (chênh lệch lớn) → dấu hiệu overfitting (mô hình học quá khớp với dữ liệu train, giảm khả năng tổng quát hóa).
- Nếu R²(train) ≈ R²(test) và cả hai đều ở mức khá → mô hình có khả năng tổng quát hóa tốt, không bị lệch nhiều.
(Với LinearRegression — mô hình tuyến tính đơn giản — hiện tượng overfitting mạnh hiếm khi xảy ra; nhiều khả năng bạn sẽ thấy 2 tập có kết quả khá gần nhau, thiên về dấu hiệu underfitting nhẹ do quan hệ giữa feature và target không hoàn toàn tuyến tính.)

#Metric nào dễ diễn giải theo đơn vị target, metric nào phạt mạnh lỗi lớn:

- MAE (Mean Absolute Error) dễ diễn giải nhất vì có cùng đơn vị với target (ở đây là 100.000 USD) — ví dụ MAE = 0.5 nghĩa là trung bình mô hình sai lệch khoảng $50,000 so với giá thực.
- RMSE cũng cùng đơn vị với target nên vẫn diễn giải được trực tiếp, nhưng phạt mạnh hơn các lỗi lớn do phép bình phương trong công thức — một vài dự đoán sai lệch rất lớn (outlier) sẽ kéo RMSE lên cao hơn nhiều so với MAE.
- MSE cũng phạt mạnh lỗi lớn (do bình phương) nhưng đơn vị bị bình phương (100.000 USD²), nên khó diễn giải trực quan, thường chỉ dùng để so sánh tương đối giữa các mô hình chứ không đọc trực tiếp theo nghĩa thực tế.
- R² không có đơn vị (là tỷ lệ %), phù hợp để đánh giá tổng thể mức độ giải thích biến thiên, nhưng không cho biết độ lớn sai số cụ thể là bao nhiêu USD.

#Câu 6. Thực nghiệm Gradient Descent và siêu tham số (10 điểm)

```python
#33 - 34
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDRegressor

pipe_base = Pipeline([
    ('scaler', StandardScaler()),
    ('sgd', SGDRegressor(max_iter=1000, learning_rate='constant', eta0=0.01, penalty=None, random_state=42))
])
pipe_base.fit(X_train, y_train)

def evaluate_config(eta0, max_iter, X_train, y_train, X_test, y_test):
    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('sgd', SGDRegressor(max_iter=max_iter, learning_rate='constant',
                              eta0=eta0, penalty=None, random_state=42))
    ])
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    return mae, rmse, r2
```

```python
#35
eta0_list = [0.0001, 0.001, 0.005]
results = []

for eta0 in eta0_list:
    mae, rmse, r2 = evaluate_config(eta0, 1000, X_train, y_train, X_test, y_test)
    results.append({'Config': f'eta0={eta0}, max_iter=1000', 'eta0': eta0, 'max_iter': 1000,
                     'MAE': mae, 'RMSE': rmse, 'R2': r2})

print(pd.DataFrame(results))
```

```python
#36
best_eta0 = pd.DataFrame(results).sort_values(by='RMSE').iloc[0]['eta0']
print("eta0 tốt nhất (theo RMSE trên test):", best_eta0)

max_iter_list = [100, 500, 2000]

for max_iter in max_iter_list:
    mae, rmse, r2 = evaluate_config(best_eta0, max_iter, X_train, y_train, X_test, y_test)
    results.append({'Config': f'eta0={best_eta0}, max_iter={max_iter}', 'eta0': best_eta0, 'max_iter': max_iter,
                     'MAE': mae, 'RMSE': rmse, 'R2': r2})
```

```python
#37
results_df = pd.DataFrame(results).drop_duplicates(subset=['eta0', 'max_iter']).reset_index(drop=True)
results_df[['MAE', 'RMSE', 'R2']] = results_df[['MAE', 'RMSE', 'R2']].round(4)
print(results_df)
```

```python
#38
plt.figure(figsize=(10, 5))
plt.bar(results_df['Config'], results_df['RMSE'], color='cornflowerblue')
plt.xticks(rotation=45, ha='right')
plt.xlabel('Cấu hình')
plt.ylabel('RMSE (tập test)')
plt.title('So sánh RMSE giữa các cấu hình SGDRegressor')
plt.tight_layout()
plt.show()
```

Nhận xét: Cấu hình có eta0 quá nhỏ thường hội tụ chậm nên RMSE cao hơn nếu max_iter không đủ lớn để bù lại; eta0 quá lớn có thể khiến quá trình học dao động, không ổn định, RMSE cũng cao. Cấu hình với eta0 vừa phải kết hợp max_iter đủ lớn thường cho RMSE thấp nhất và ổn định nhất trong biểu đồ.

#Nhận xét:
Ảnh hưởng của learning rate (eta0):

- Quá nhỏ: mỗi bước cập nhật trọng số rất nhỏ → mô hình hội tụ rất chậm, nếu max_iter không đủ lớn thì thuật toán dừng trước khi đạt điểm tối ưu, dẫn đến underfitting tạm thời (chưa học đủ, sai số cao dù bản chất mô hình có thể học tốt hơn nếu chạy đủ lâu).
- Quá lớn: bước cập nhật quá lớn khiến quá trình tối ưu dao động qua lại quanh điểm tối ưu hoặc phân kỳ (loss không giảm ổn định, thậm chí tăng), khiến mô hình không hội tụ được, cho kết quả kém và không ổn định giữa các lần chạy.
Learning rate phù hợp giúp mô hình hội tụ đủ nhanh mà vẫn ổn định, tiệm cận điểm tối ưu của hàm mất mát.

Ảnh hưởng của max_iter quá nhỏ:
- Nếu max_iter quá nhỏ, thuật toán dừng lại trước khi hội tụ (chưa đạt đến điểm tối ưu của hàm mất mát), dẫn đến mô hình chưa học đủ — biểu hiện tương tự underfitting: cả sai số trên train và test đều cao hơn mức mô hình có thể đạt được.

Về việc chọn cấu hình "tốt nhất":
- Việc chọn cấu hình tốt nhất trong bài này dựa trên RMSE/R² đo trên tập test, không phải trên tập train — vì điểm số cao trên train có thể chỉ phản ánh mô hình khớp tốt với dữ liệu đã thấy (có nguy cơ overfitting), không đảm bảo khả năng tổng quát hóa. Đây là lý do vì sao đề yêu cầu không chọn "tốt nhất" chỉ dựa trên train score — phải luôn đánh giá trên tập test (dữ liệu mô hình chưa từng thấy) để có kết luận đáng tin cậy.

# 44. Vì sao MAE và RMSE có thể dẫn đến đánh giá khác nhau? Khi nào ưu tiên RMSE?

- MAE tính trung bình trị tuyệt đối của sai số, coi mọi lỗi có trọng số như nhau; RMSE bình phương sai số trước khi lấy trung bình rồi khai căn, nên phạt nặng các lỗi lớn hơn nhiều so với lỗi nhỏ.
- Trong kết quả của nhóm, LinearRegression có MAE = 0.5332 nhưng RMSE = 0.7456 — chênh lệch khá lớn giữa hai chỉ số này chính là dấu hiệu cho thấy tồn tại một số sai số cực lớn (như mẫu index 1979 với Absolute Error = 9.875), khiến RMSE bị kéo lên cao hơn nhiều so với mức "trung bình thông thường" mà MAE phản ánh. - Nếu hai mô hình có MAE giống nhau nhưng RMSE khác nhau, mô hình có RMSE thấp hơn là mô hình có ít lỗi cực đoan hơn dù mức sai số điển hình tương đương.
- Nên ưu tiên RMSE khi bài toán coi lỗi lớn là đặc biệt tốn kém hoặc nguy hiểm (ví dụ định giá sai nghiêm trọng một bất động sản gây thiệt hại tài chính lớn hơn nhiều so với sai số nhỏ dàn trải).

Đây là chỉ số đánh giá tổng thể trên toàn bộ tập dữ liệu, không phải cam kết cho từng mẫu riêng lẻ.

#46. Vì sao LinearRegression gần như không cần learning rate, còn SGDRegressor cần?

- LinearRegression (mặc định trong sklearn) giải bài toán bằng phương pháp đại số closed-form (Normal Equation/OLS) — tính trực tiếp nghiệm tối ưu bằng công thức ma trận mà không cần lặp dần, nên không có khái niệm "bước cập nhật" và do đó không cần learning rate. SGDRegressor ngược lại tối ưu bằng gradient descent lặp, cập nhật trọng số từng bước nhỏ theo hướng giảm gradient của hàm mất mát, và learning rate (eta0) chính là kích thước mỗi bước đó.
- Kết quả thực nghiệm của nhóm minh họa rõ điều này: khi eta0 quá lớn (0.001, 0.005), SGDRegressor phân kỳ hoàn toàn (R² âm tới -4.9×10¹⁹), trong khi LinearRegression không hề gặp vấn đề tương tự vì nó không phụ thuộc vào bất kỳ tham số bước lặp nào.

#47. Nếu số mẫu tăng gấp 10 lần, phần nào tốn tài nguyên hơn và cần thay đổi gì?

- Với LinearRegression (OLS, giải bằng đại số ma trận), chi phí tính toán tăng theo cả số mẫu và đặc biệt số chiều đặc trưng, nhưng vẫn khả thi vì bài toán chỉ có 8 feature; phần tốn tài nguyên hơn rõ rệt sẽ là các bước EDA (vẽ scatter/histogram trên toàn bộ dữ liệu, tính ma trận tương quan, boxplot) — với 10 lần dữ liệu, cần tăng cường lấy mẫu (sampling) khi vẽ biểu đồ (như nhóm đã làm với df.sample(n=4000)) để tránh biểu đồ quá dày đặc và chậm render.
- Với SGDRegressor, số lượng vòng lặp cần thiết để hội tụ có thể ít thay đổi vì đây là thuật toán mini-batch/incremental, phù hợp hơn khi dữ liệu lớn — đây chính là lý do SGD thường được ưu tiên hơn OLS trong các bài toán dữ liệu khổng lồ. Ngoài ra, việc chia train/test và tính toán các metric (MAE, RMSE, R²) trên tập test cũng cần nhiều bộ nhớ hơn, nên cần cân nhắc xử lý theo batch nếu dữ liệu vượt quá khả năng RAM.

```python
from sklearn.ensemble import RandomForestRegressor
import time

rf_model = RandomForestRegressor(random_state=42, n_jobs=-1)

start = time.time()
rf_model.fit(X_train, y_train)
rf_train_time = time.time() - start

y_train_pred_rf = rf_model.predict(X_train)
y_test_pred_rf = rf_model.predict(X_test)

mae_train_rf = mean_absolute_error(y_train, y_train_pred_rf)
rmse_train_rf = np.sqrt(mean_squared_error(y_train, y_train_pred_rf))
r2_train_rf = r2_score(y_train, y_train_pred_rf)

mae_test_rf = mean_absolute_error(y_test, y_test_pred_rf)
rmse_test_rf = np.sqrt(mean_squared_error(y_test, y_test_pred_rf))
r2_test_rf = r2_score(y_test, y_test_pred_rf)

print(f"Thời gian huấn luyện RandomForest: {rf_train_time:.2f} giây")
print("Train - MAE:", mae_train_rf, "RMSE:", rmse_train_rf, "R2:", r2_train_rf)
print("Test  - MAE:", mae_test_rf, "RMSE:", rmse_test_rf, "R2:", r2_test_rf)
```
